```python
logins = [
    ["alice", "success", "10:15"],
    ["bob", "failed", "02:10"],
    ["bob", "failed", "02:11"],
    ["bob", "failed", "02:12"],
    ["carol", "success", "14:30"],
    ["alice", "failed", "16:45"],
]

failed_count = {}
suspicious = []
total_failed = 0

for user, status, time in logins:
    hour = int(time.split(":")[0])

    if status == "failed":
        total_failed = total_failed + 1
        failed_count[user] = failed_count.get(user, 0) + 1

    if hour < 5:
        suspicious.append(user)

for user, count in failed_count.items():
    if count >= 3:
        print("ALERT:", user, "had", count, "failed logins")

print("Total logins:", len(logins))
print("Total failed:", total_failed)
print("Logins between 00:00 and 05:00:", suspicious)
```
